```python
from __future__ import annotations

import argparse
import contextlib
import logging
from itertools import islice
import gc
import html
import json
import math
import os
import re
import shutil
import sys
import unicodedata
import zipfile
from collections import Counter, defaultdict
from dataclasses import asdict, dataclass
from datetime import datetime
from pathlib import Path
from typing import Any, Iterable

import cv2
import numpy as np
import pandas as pd
from PIL import Image, ImageOps
from PIL.ExifTags import GPSTAGS, TAGS


IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".webp"}


def file_path(value: str | Path) -> Path:
    """Support long OneDrive paths on Windows."""
    path = str(Path(value).absolute())
    if os.name == "nt" and not path.startswith("\\\\?\\"):
        path = "\\\\?\\UNC\\" + path[2:] if path.startswith("\\\\") else "\\\\?\\" + path
    return Path(path)


@dataclass(frozen=True)
class TargetDefinition:
    label: str
    prompt: str
    group: str
    priority: str
    color: str
    action: str
    synonyms: tuple[str, ...] = ()
    internal: bool = False


@dataclass(frozen=True)
class SearchPass:
    name: str
    kind: str
    max_side: int = 1600
    tile_size: int = 960
    overlap: int = 160


@dataclass(frozen=True)
class SearchCrop:
    image: np.ndarray
    x0: int
    y0: int
    scale: float
    pass_name: str


@dataclass
class Detection:
    image: str
    target: str
    prompt: str
    group: str
    priority: str
    confidence: float
    x1: int
    y1: int
    x2: int
    y2: int
    width_px: int
    height_px: int
    area_ratio: float
    source_pass: str
    gps_lat: float | None = None
    gps_lon: float | None = None
    gps_alt_m: float | None = None
    relative_alt_m: float | None = None
    focal_length_mm: float | None = None
    focal_length_35mm: float | None = None
    taken_at: str | None = None
    is_internal: bool = False


@dataclass
class Alert:
    image: str
    tipo: str
    severidad: str
    confianza: float
    detalle: str
    accion_sugerida: str
    x1: int | None = None
    y1: int | None = None
    x2: int | None = None
    y2: int | None = None
    evidencia: str | None = None


@dataclass
class ScaleEstimate:
    image: str
    m_per_px: float | None
    source: str
    confidence: str
    label: str
    bar_m: float | None = None
    bar_px: int | None = None


@dataclass
class ImageSummary:
    image: str
    status: str
    total_detections: int
    findings: str
    original_width: int
    original_height: int
    processed_width: int
    processed_height: int
    annotated_path: str | None = None
    gps_lat: float | None = None
    gps_lon: float | None = None
    gps_alt_m: float | None = None
    relative_alt_m: float | None = None
    focal_length_mm: float | None = None
    focal_length_35mm: float | None = None
    scale_m_per_px: float | None = None
    scale_source: str | None = None
    alert_count: int = 0
    taken_at: str | None = None


COMMON_ACTIONS = {
    "EPP": "Verificar dotacion, uso y registro de EPP antes de liberar el frente.",
    "Senalizacion": "Revisar delimitacion, carteleria visible y continuidad del vallado.",
    "Vehiculos": "Separar circulacion peatonal/equipos y confirmar condiciones de maniobra.",
    "Maquinaria": "Validar estado operativo, radios de giro y distancia segura con personal.",
    "Ductos": "Contrastar contra traza, interferencias, canerias, mojones y puntos de control.",
    "Excavaciones": "Revisar accesos, estabilidad, vallado, distancia a bordes y servicios existentes.",
    "Altura": "Confirmar plataforma, linea de vida, barandas y planilla de control de equipos.",
    "Orden y limpieza": "Registrar condicion, retirar obstaculos y normalizar acopios o residuos.",
}


CATALOG = (
    TargetDefinition(
        "Cono vial",
        "traffic cone",
        "Senalizacion",
        "Media",
        "#f97316",
        COMMON_ACTIONS["Senalizacion"],
        ("cono", "conos", "cono vial", "traffic cone", "baliza"),
    ),
    TargetDefinition(
        "Cartel de seguridad",
        "safety sign",
        "Senalizacion",
        "Alta",
        "#eab308",
        COMMON_ACTIONS["Senalizacion"],
        ("cartel", "carteles", "senal", "senalizacion", "safety sign", "warning sign"),
    ),
    TargetDefinition(
        "Casco",
        "hard hat",
        "EPP",
        "Alta",
        "#22c55e",
        COMMON_ACTIONS["EPP"],
        ("casco", "cascos", "hardhat", "hard hat", "helmet", "casco de seguridad"),
    ),
    TargetDefinition(
        "Chaleco reflectivo",
        "safety vest",
        "EPP",
        "Alta",
        "#84cc16",
        COMMON_ACTIONS["EPP"],
        ("chaleco", "chalecos", "chaleco reflectivo", "safety vest", "reflective vest"),
    ),
    TargetDefinition(
        "Persona",
        "person",
        "EPP",
        "Media",
        "#14b8a6",
        "Cruzar presencia de personal con EPP visible y segregacion de equipos.",
        ("persona", "personas", "operario", "trabajador", "worker", "person"),
    ),
    TargetDefinition(
        "Auto",
        "car",
        "Vehiculos",
        "Media",
        "#2563eb",
        COMMON_ACTIONS["Vehiculos"],
        ("auto", "autos", "coche", "car", "cars", "vehiculo liviano"),
    ),
    TargetDefinition(
        "Camioneta",
        "pickup truck",
        "Vehiculos",
        "Media",
        "#1d4ed8",
        COMMON_ACTIONS["Vehiculos"],
        ("camioneta", "pickup", "pickup truck", "4x4", "utilitario"),
    ),
    TargetDefinition(
        "Camion",
        "truck",
        "Vehiculos",
        "Media",
        "#0f766e",
        COMMON_ACTIONS["Vehiculos"],
        ("camion", "camiones", "truck", "dump truck", "volcador"),
    ),
    TargetDefinition(
        "Excavadora",
        "excavator",
        "Maquinaria",
        "Alta",
        "#a855f7",
        COMMON_ACTIONS["Maquinaria"],
        ("excavadora", "retroexcavadora", "excavator", "backhoe"),
    ),
    TargetDefinition(
        "Grua",
        "crane",
        "Maquinaria",
        "Alta",
        "#7c3aed",
        COMMON_ACTIONS["Maquinaria"],
        ("grua", "crane", "hidrogrua", "sideboom"),
    ),
    TargetDefinition(
        "Caneria/Tuberia",
        "pipe",
        "Ductos",
        "Media",
        "#0891b2",
        COMMON_ACTIONS["Ductos"],
        ("caneria", "canerias", "tuberia", "tuberias", "pipe", "pipeline", "ducto"),
    ),
    TargetDefinition(
        "Zanja abierta",
        "trench",
        "Excavaciones",
        "Alta",
        "#b45309",
        COMMON_ACTIONS["Excavaciones"],
        ("zanja", "zanjas", "excavacion", "excavaciones", "trench", "ditch"),
    ),
    TargetDefinition(
        "Mojon/estaca",
        "survey stake",
        "Ductos",
        "Media",
        "#dc2626",
        COMMON_ACTIONS["Ductos"],
        ("mojon", "mojones", "estaca", "estacas", "survey stake", "survey marker"),
    ),
    TargetDefinition(
        "Escalera",
        "ladder",
        "Altura",
        "Alta",
        "#f43f5e",
        COMMON_ACTIONS["Altura"],
        ("escalera", "escaleras", "ladder"),
    ),
    TargetDefinition(
        "Andamio",
        "scaffolding",
        "Altura",
        "Alta",
        "#be123c",
        COMMON_ACTIONS["Altura"],
        ("andamio", "andamios", "scaffold", "scaffolding"),
    ),
    TargetDefinition(
        "Residuo u obstaculo",
        "construction debris",
        "Orden y limpieza",
        "Media",
        "#64748b",
        COMMON_ACTIONS["Orden y limpieza"],
        ("residuo", "residuos", "escombro", "escombros", "obstaculo", "debris"),
    ),
)


SAFETY_INTERNAL_TARGETS = (
    TargetDefinition(
        "Casco",
        "safety helmet",
        "Control SyH",
        "Alta",
        "#22c55e",
        COMMON_ACTIONS["EPP"],
        ("safety helmet", "construction helmet"),
        True,
    ),
    TargetDefinition(
        "Telefono celular",
        "cell phone",
        "Control actividad",
        "Media",
        "#ef4444",
        "Validar si la persona esta usando telefono o se encuentra fuera de tarea.",
        ("telefono", "celular", "mobile phone", "smartphone"),
        True,
    ),
    TargetDefinition(
        "Persona sentada",
        "sitting person",
        "Control actividad",
        "Media",
        "#ef4444",
        "Validar si la persona esta en descanso autorizado o fuera de tarea.",
        ("persona sentada", "sitting worker", "person sitting"),
        True,
    ),
    TargetDefinition(
        "Persona recostada",
        "lying person",
        "Control actividad",
        "Media",
        "#ef4444",
        "Validar si existe una condicion insegura, descanso no autorizado o emergencia.",
        ("persona acostada", "lying worker", "person lying"),
        True,
    ),
    TargetDefinition(
        "Silla",
        "chair",
        "Control actividad",
        "Baja",
        "#64748b",
        "Usar como indicio contextual; validar visualmente.",
        ("silla", "chair"),
        True,
    ),
)


DEFAULT_PRESET_LABELS = (
    "Cono vial",
    "Cartel de seguridad",
    "Casco",
    "Chaleco reflectivo",
    "Persona",
    "Auto",
    "Camioneta",
    "Camion",
    "Excavadora",
    "Caneria/Tuberia",
    "Zanja abierta",
    "Mojon/estaca",
)


SEARCH_PROFILES = {
    "rapida": (
        SearchPass("general_1280", "global", max_side=1280),
    ),
    "equilibrada": (
        SearchPass("general_1600", "global", max_side=1600),
        SearchPass("tiles_960", "tile", tile_size=960, overlap=180),
    ),
    "profunda": (
        SearchPass("general_1800", "global", max_side=1800),
        SearchPass("tiles_960", "tile", tile_size=960, overlap=220),
        SearchPass("tiles_640", "tile", tile_size=640, overlap=180),
    ),
}


def normalize_key(value: str) -> str:
    value = unicodedata.normalize("NFKD", str(value))
    value = value.encode("ascii", "ignore").decode("ascii")
    value = re.sub(r"[^a-zA-Z0-9]+", " ", value.lower()).strip()
    return re.sub(r"\s+", " ", value)


def split_terms(value: str | Iterable[str] | None) -> list[str]:
    if value is None:
        return []
    if isinstance(value, str):
        parts = re.split(r"[,;\n]+", value)
    else:
        parts = list(value)
    return [str(part).strip() for part in parts if str(part).strip()]


def catalog_lookup() -> dict[str, TargetDefinition]:
    lookup: dict[str, TargetDefinition] = {}
    for target in CATALOG:
        keys = (target.label, target.prompt, *target.synonyms)
        for key in keys:
            lookup[normalize_key(key)] = target
    return lookup


def build_targets(
    preset_labels: Iterable[str] | None = None,
    extra_terms: str | Iterable[str] | None = None,
) -> list[TargetDefinition]:
    lookup = catalog_lookup()
    selected: list[TargetDefinition] = []

    for label in preset_labels or ():
        target = lookup.get(normalize_key(label))
        if target is not None:
            selected.append(target)

    for term in split_terms(extra_terms):
        target = lookup.get(normalize_key(term))
        if target is None:
            clean = term.strip()
            target = TargetDefinition(
                label=clean.title(),
                prompt=clean,
                group="Busqueda personalizada",
                priority="Media",
                color="#475569",
                action="Validar visualmente el hallazgo y clasificarlo en el parte tecnico.",
                synonyms=(clean,),
            )
        selected.append(target)

    deduped: dict[str, TargetDefinition] = {}
    for target in selected:
        deduped[normalize_key(target.prompt)] = target
    return list(deduped.values())


def ensure_safety_control_targets(targets: Iterable[TargetDefinition], activity_cues: bool = False) -> list[TargetDefinition]:
    lookup = catalog_lookup()
    merged: dict[str, TargetDefinition] = {normalize_key(target.prompt): target for target in targets}

    for required in ("persona", "casco"):
        target = lookup.get(required)
        if target is not None:
            merged.setdefault(normalize_key(target.prompt), target)

    for target in SAFETY_INTERNAL_TARGETS:
        if target.group == "Control actividad" and not activity_cues:
            continue
        merged.setdefault(normalize_key(target.prompt), target)

    return list(merged.values())


def image_files(input_folder: str | Path) -> list[Path]:
    folder = file_path(input_folder)
    files = [path for path in folder.iterdir() if path.is_file() and path.suffix.lower() in IMAGE_EXTENSIONS]
    return sorted(files, key=lambda path: path.name.lower())


def hex_to_bgr(hex_color: str) -> tuple[int, int, int]:
    cleaned = hex_color.strip().lstrip("#")
    if len(cleaned) != 6:
        return (40, 120, 220)
    r = int(cleaned[0:2], 16)
    g = int(cleaned[2:4], 16)
    b = int(cleaned[4:6], 16)
    return (b, g, r)


def rational_to_float(value: Any) -> float | None:
    try:
        if isinstance(value, tuple) and len(value) == 2:
            return float(value[0]) / float(value[1])
        return float(value)
    except Exception:
        return None


def extract_xmp_number(path: str | Path, names: Iterable[str]) -> float | None:
    try:
        with open(path, "rb") as fh:
            raw = fh.read(768_000)
        text = raw.decode("utf-8", errors="ignore")
    except Exception:
        return None

    for name in names:
        patterns = (
            rf'{re.escape(name)}\s*=\s*"([+-]?\d+(?:\.\d+)?)"',
            rf"<[^>]*{re.escape(name)}[^>]*>\s*([+-]?\d+(?:\.\d+)?)\s*</",
        )
        for pattern in patterns:
            match = re.search(pattern, text, flags=re.IGNORECASE)
            if match:
                return rational_to_float(match.group(1))
    return None


def gps_coord_to_decimal(values: Any, ref: str | None) -> float | None:
    if not values or len(values) < 3:
        return None
    degrees = rational_to_float(values[0])
    minutes = rational_to_float(values[1])
    seconds = rational_to_float(values[2])
    if degrees is None or minutes is None or seconds is None:
        return None
    result = degrees + minutes / 60 + seconds / 3600
    if ref in {"S", "W"}:
        result *= -1
    return result


def extract_exif_metadata(path: str | Path) -> dict[str, Any]:
    metadata: dict[str, Any] = {
        "gps_lat": None,
        "gps_lon": None,
        "gps_alt_m": None,
        "relative_alt_m": extract_xmp_number(path, ("RelativeAltitude", "drone-dji:RelativeAltitude")),
        "focal_length_mm": None,
        "focal_length_35mm": None,
        "taken_at": None,
    }
    try:
        with Image.open(path) as image:
            exif = image.getexif()
            if not exif:
                return metadata
            decoded = {TAGS.get(tag, tag): value for tag, value in exif.items()}
            taken_at = decoded.get("DateTimeOriginal") or decoded.get("DateTime")
            if taken_at:
                metadata["taken_at"] = str(taken_at)
            decoded.update({TAGS.get(tag, tag): value for tag, value in exif.get_ifd(34665).items()} if 34665 in exif else {})
            metadata["focal_length_mm"] = rational_to_float(decoded.get("FocalLength"))
            metadata["focal_length_35mm"] = rational_to_float(decoded.get("FocalLengthIn35mmFilm"))

            gps_info = exif.get_ifd(34853) if 34853 in exif else None
            if gps_info:
                gps = {GPSTAGS.get(key, key): value for key, value in gps_info.items()}
                metadata["gps_lat"] = gps_coord_to_decimal(gps.get("GPSLatitude"), gps.get("GPSLatitudeRef"))
                metadata["gps_lon"] = gps_coord_to_decimal(gps.get("GPSLongitude"), gps.get("GPSLongitudeRef"))
                altitude = rational_to_float(gps.get("GPSAltitude"))
                if altitude is not None and gps.get("GPSAltitudeRef") == 1:
                    altitude *= -1
                metadata["gps_alt_m"] = altitude
    except Exception:
        return metadata
    return metadata


def read_image_bgr(path: str | Path) -> tuple[np.ndarray, int, int]:
    with Image.open(path) as pil_image:
        pil_image = ImageOps.exif_transpose(pil_image).convert("RGB")
        width, height = pil_image.size
        rgb = np.array(pil_image)
    return cv2.cvtColor(rgb, cv2.COLOR_RGB2BGR), width, height


def resize_to_max_side(image: np.ndarray, max_side: int) -> tuple[np.ndarray, float]:
    height, width = image.shape[:2]
    longest = max(height, width)
    if max_side == 0 or longest <= max_side:
        return image, 1.0
    scale = max_side / float(longest)
    resized = cv2.resize(image, (int(width * scale), int(height * scale)), interpolation=cv2.INTER_AREA)
    return resized, scale


def starts_for_tiles(length: int, tile_size: int, overlap: int) -> list[int]:
    if length <= 0 or tile_size <= 0 or not 0 <= overlap < tile_size:
        raise ValueError("Dimensiones positivas y 0 <= overlap < tile_size requeridos.")
    if length <= tile_size:
        return [0]
    step = max(1, tile_size - overlap)
    starts = list(range(0, length - tile_size + 1, step))
    last = max(0, length - tile_size)
    if not starts or starts[-1] != last:
        starts.append(last)
    return sorted(set(starts))


def build_search_crops(image: np.ndarray, passes: Iterable[SearchPass]) -> Iterable[SearchCrop]:
    """Yield native-resolution views; retain only the current inference batch."""
    height, width = image.shape[:2]
    full_seen = False
    for search_pass in passes:
        if search_pass.kind == "global":
            resized, scale = resize_to_max_side(image, search_pass.max_side)
            full_seen |= scale == 1.0
            yield SearchCrop(resized, 0, 0, scale, search_pass.name)
        elif search_pass.kind == "tile":
            tile = search_pass.tile_size
            if height <= tile and width <= tile and full_seen:
                continue
            for y0 in starts_for_tiles(height, tile, search_pass.overlap):
                for x0 in starts_for_tiles(width, tile, search_pass.overlap):
                    yield SearchCrop(image[y0:y0+tile, x0:x0+tile], x0, y0, 1.0, search_pass.name)
            full_seen |= height <= tile and width <= tile
        else:
            raise ValueError(f"Tipo de pasada desconocido: {search_pass.kind}")


def safe_float(value: Any) -> float:
    try:
        if hasattr(value, "detach"):
            value = value.detach().cpu().numpy()
        if isinstance(value, np.ndarray):
            return float(value.reshape(-1)[0])
        return float(value)
    except Exception:
        return 0.0


def safe_xyxy(value: Any) -> tuple[float, float, float, float]:
    if hasattr(value, "detach"):
        value = value.detach().cpu().numpy()
    array = np.array(value).reshape(-1)
    return tuple(float(v) for v in array[:4])  # type: ignore[return-value]


def clamp_box(x1: float, y1: float, x2: float, y2: float, width: int, height: int) -> tuple[int, int, int, int]:
    ix1 = max(0, min(width - 1, int(round(x1))))
    iy1 = max(0, min(height - 1, int(round(y1))))
    ix2 = max(0, min(width, int(round(x2))))
    iy2 = max(0, min(height, int(round(y2))))
    if ix2 <= ix1:
        ix2 = min(width, ix1 + 1)
    if iy2 <= iy1:
        iy2 = min(height, iy1 + 1)
    return ix1, iy1, ix2, iy2


def box_area(box: tuple[int, int, int, int]) -> int:
    return max(0, box[2] - box[0]) * max(0, box[3] - box[1])


def detection_box(det: Detection) -> tuple[int, int, int, int]:
    return det.x1, det.y1, det.x2, det.y2


def expand_box(
    box: tuple[int, int, int, int],
    pad_x: float,
    pad_y: float,
    width: int,
    height: int,
) -> tuple[int, int, int, int]:
    x1, y1, x2, y2 = box
    bw = x2 - x1
    bh = y2 - y1
    return clamp_box(x1 - bw * pad_x, y1 - bh * pad_y, x2 + bw * pad_x, y2 + bh * pad_y, width, height)


def intersection_area(a: tuple[int, int, int, int], b: tuple[int, int, int, int]) -> int:
    x1 = max(a[0], b[0])
    y1 = max(a[1], b[1])
    x2 = min(a[2], b[2])
    y2 = min(a[3], b[3])
    return box_area((x1, y1, x2, y2))


def iou(a: tuple[int, int, int, int], b: tuple[int, int, int, int]) -> float:
    inter = intersection_area(a, b)
    union = box_area(a) + box_area(b) - inter
    return inter / union if union else 0.0


def center_inside(inner: Detection, outer: tuple[int, int, int, int]) -> bool:
    cx = (inner.x1 + inner.x2) / 2
    cy = (inner.y1 + inner.y2) / 2
    return outer[0] <= cx <= outer[2] and outer[1] <= cy <= outer[3]


def nice_distance(value_m: float) -> float:
    if value_m <= 0:
        return 0.0
    exponent = math.floor(math.log10(value_m))
    base = value_m / (10**exponent)
    if base < 1.5:
        nice = 1
    elif base < 3.5:
        nice = 2
    elif base < 7.5:
        nice = 5
    else:
        nice = 10
    return nice * (10**exponent)


class VisualInspectionTool:
    def __init__(
        self,
        input_folder: str | Path,
        output_folder: str | Path = "output_reporte_visual",
        targets: Iterable[TargetDefinition] | None = None,
        confidence: float = 0.20,
        search_profile: str = "equilibrada",
        nms_iou: float = 0.45,
        max_image_side: int = 0,
        batch_size: int = 8,
        imgsz: int = 960,
        model_name: str = "yolov8s-worldv2.pt",
        safety_control: bool = True,
        activity_cues: bool = False,
        device: str | None = None,
    ) -> None:
        self.input_folder = file_path(input_folder)
        self.output_folder = file_path(output_folder) / datetime.now().strftime("industrial_%Y%m%d_%H%M%S_%f")
        self.annotated_folder = self.output_folder / "imagenes_marcadas"
        self.tables_folder = self.output_folder / "tablas"
        self.output_folder.mkdir(parents=True, exist_ok=True)
        self.annotated_folder.mkdir(parents=True, exist_ok=True)
        self.tables_folder.mkdir(parents=True, exist_ok=True)

        self.safety_control = safety_control
        self.activity_cues = activity_cues
        self.device = device
        selected = list(targets) if targets is not None else build_targets(DEFAULT_PRESET_LABELS)
        self.targets = ensure_safety_control_targets(selected, activity_cues) if safety_control else selected
        if not self.targets:
            raise ValueError("No hay objetivos de busqueda configurados.")

        self.prompts = [target.prompt for target in self.targets]
        self.prompt_to_target = {normalize_key(target.prompt): target for target in self.targets}
        if not 0 < confidence <= 1 or not 0 <= nms_iou <= 1:
            raise ValueError("Confianza debe ser (0, 1] e IoU [0, 1].")
        if batch_size < 1 or imgsz < 32 or max_image_side < 0:
            raise ValueError("Lote >= 1, imgsz >= 32 y max_image_side >= 0.")
        if search_profile not in SEARCH_PROFILES:
            raise ValueError(f"Perfil desconocido: {search_profile}")
        self.confidence = float(confidence)
        self.search_profile = search_profile if search_profile in SEARCH_PROFILES else "equilibrada"
        self.search_passes = SEARCH_PROFILES[self.search_profile]
        self.nms_iou = float(nms_iou)
        self.max_image_side = int(max_image_side)
        self.batch_size = max(1, int(batch_size))
        self.imgsz = int(imgsz)
        self.model_name = model_name
        self.model = None
        self.detections: list[Detection] = []
        self.alerts: list[Alert] = []
        self.image_summaries: list[ImageSummary] = []
        self.started_at = datetime.now()
        self.finished_at: datetime | None = None

    def load_model(self) -> None:
        if self.model is not None:
            return
        try:
            from ultralytics import YOLOWorld
        except ImportError as exc:
            raise RuntimeError(
                "Falta ultralytics. En Colab ejecuta la celda de instalacion; localmente usa: "
                "pip install ultralytics opencv-python pillow reportlab pandas openpyxl"
            ) from exc

        self.model = YOLOWorld(self.model_name)
        self.model.set_classes(self.prompts)

    def process(self) -> list[ImageSummary]:
        if self.image_summaries or self.finished_at is not None:
            self.output_folder = self.output_folder.parent / datetime.now().strftime("industrial_%Y%m%d_%H%M%S_%f")
            self.annotated_folder = self.output_folder / "imagenes_marcadas"
            self.tables_folder = self.output_folder / "tablas"
            self.annotated_folder.mkdir(parents=True)
            self.tables_folder.mkdir(parents=True)
        self.detections.clear()
        self.alerts.clear()
        self.image_summaries.clear()
        self.finished_at = None
        self.started_at = datetime.now()
        files = image_files(self.input_folder)
        if not files:
            raise FileNotFoundError(f"No se encontraron imagenes en {self.input_folder}")

        self.load_model()
        print(f"Imagenes a procesar: {len(files)}")
        print(f"Objetivos simultaneos: {', '.join(target.label for target in self.targets)}")
        print(f"Perfil de busqueda: {self.search_profile} ({len(self.search_passes)} pasada/s)")

        for index, image_path in enumerate(files, start=1):
            print(f"[{index}/{len(files)}] {image_path.name}")
            summary = self.process_image(image_path)
            self.image_summaries.append(summary)

        self.finished_at = datetime.now()
        return self.image_summaries

    def process_image(self, image_path: Path) -> ImageSummary:
        metadata = extract_exif_metadata(image_path)
        image, original_width, original_height = read_image_bgr(image_path)
        image, process_scale = resize_to_max_side(image, self.max_image_side)
        processed_height, processed_width = image.shape[:2]
        crops = build_search_crops(image, self.search_passes)
        raw_detections = self.predict_crops(crops, image_path.name, processed_width, processed_height, metadata)
        detections = self.apply_nms(raw_detections)
        scale = self.estimate_scale(image_path.name, detections, processed_width, processed_height, metadata)
        alerts = self.evaluate_alerts(image_path.name, detections, processed_width, processed_height) if self.safety_control else []

        if process_scale != 1.0:
            note = f"procesada a {processed_width}x{processed_height}px para optimizar memoria"
        else:
            note = f"procesada a resolucion original {processed_width}x{processed_height}px"
        print(f"  sectores: {self.last_crop_count} | detecciones: {len(detections)} | alertas: {len(alerts)} | {note}")

        self.detections.extend(detections)
        self.alerts.extend(alerts)
        counts = Counter(det.target for det in detections if not det.is_internal)
        findings = "; ".join(f"{label}: {count}" for label, count in counts.most_common()) or "Sin hallazgos"
        annotated = self.draw_detections(image.copy(), image_path.name, detections, alerts, scale)
        annotated_path = str(self.annotated_folder / f"{image_path.name}_marcada.jpg")
        ok, encoded = cv2.imencode(".jpg", annotated, [int(cv2.IMWRITE_JPEG_QUALITY), 92])
        if not ok:
            raise OSError(f"No se pudo codificar {annotated_path}")
        encoded.tofile(annotated_path)

        return ImageSummary(
            image=image_path.name,
            status="Con alertas" if alerts else ("Con hallazgos" if counts else "Sin hallazgos"),
            total_detections=sum(1 for det in detections if not det.is_internal),
            findings=findings,
            original_width=original_width,
            original_height=original_height,
            processed_width=processed_width,
            processed_height=processed_height,
            annotated_path=annotated_path,
            scale_m_per_px=scale.m_per_px,
            scale_source=scale.source,
            alert_count=len(alerts),
            **metadata,
        )

    def predict_crops(
        self,
        crops: Iterable[SearchCrop],
        image_name: str,
        image_width: int,
        image_height: int,
        metadata: dict[str, Any],
    ) -> list[Detection]:
        detections: list[Detection] = []
        if self.model is None:
            raise RuntimeError("El modelo no fue cargado.")

        iterator = iter(crops)
        self.last_crop_count = 0
        while batch := list(islice(iterator, self.batch_size)):
            self.last_crop_count += len(batch)
            images = [crop.image for crop in batch]
            results = self.predict_batch(images)
            if len(results) != len(batch):
                raise RuntimeError("El modelo no devolvio un resultado por sector.")
            for crop, result in zip(batch, results):
                boxes = getattr(result, "boxes", [])
                for box in boxes:
                    class_id = int(safe_float(box.cls[0]))
                    confidence = safe_float(box.conf[0])
                    if confidence < self.confidence or class_id < 0 or class_id >= len(self.prompts):
                        continue

                    prompt = self.prompts[class_id]
                    target = self.prompt_to_target.get(normalize_key(prompt))
                    if target is None:
                        continue

                    bx1, by1, bx2, by2 = safe_xyxy(box.xyxy[0])
                    if not np.isfinite([bx1, by1, bx2, by2, confidence]).all() or bx2 <= bx1 or by2 <= by1:
                        continue
                    x1 = bx1 / crop.scale + crop.x0
                    y1 = by1 / crop.scale + crop.y0
                    x2 = bx2 / crop.scale + crop.x0
                    y2 = by2 / crop.scale + crop.y0
                    ix1, iy1, ix2, iy2 = clamp_box(x1, y1, x2, y2, image_width, image_height)
                    box_width = ix2 - ix1
                    box_height = iy2 - iy1
                    area_ratio = (box_width * box_height) / max(1, image_width * image_height)
                    detections.append(
                        Detection(
                            image=image_name,
                            target=target.label,
                            prompt=prompt,
                            group=target.group,
                            priority=target.priority,
                            confidence=confidence,
                            x1=ix1,
                            y1=iy1,
                            x2=ix2,
                            y2=iy2,
                            width_px=box_width,
                            height_px=box_height,
                            area_ratio=area_ratio,
                            source_pass=crop.pass_name,
                            is_internal=target.internal and all(t.internal for t in self.targets if t.label == target.label),
                            **metadata,
                        )
                    )
        return detections

    def predict_batch(self, images: list[np.ndarray]) -> list[Any]:
        if self.model is None:
            raise RuntimeError("El modelo no fue cargado.")

        import torch

        try:
            with torch.inference_mode():
                return list(self.model.predict(
                    source=images, conf=self.confidence, imgsz=self.imgsz,
                    batch=len(images), device=self.device, verbose=False,
                ))
        except torch.cuda.OutOfMemoryError:
            if len(images) <= 1:
                raise
        # Retry only an actual CUDA memory failure, after releasing its traceback.
        torch.cuda.empty_cache()
        middle = len(images) // 2
        logging.warning("Memoria GPU insuficiente: dividiendo lote de %s", len(images))
        return self.predict_batch(images[:middle]) + self.predict_batch(images[middle:])

    def apply_nms(self, detections: list[Detection]) -> list[Detection]:
        if not detections:
            return []

        by_prompt: dict[str, list[Detection]] = defaultdict(list)
        for detection in detections:
            by_prompt[detection.target].append(detection)

        final: list[Detection] = []
        for prompt, prompt_detections in by_prompt.items():
            boxes = [
                [det.x1, det.y1, max(1, det.x2 - det.x1), max(1, det.y2 - det.y1)]
                for det in prompt_detections
            ]
            scores = [det.confidence for det in prompt_detections]
            indices = cv2.dnn.NMSBoxes(boxes, scores, score_threshold=0.0, nms_threshold=self.nms_iou)
            if indices is None or len(indices) == 0:
                continue
            flat_indices = np.array(indices).reshape(-1)
            final.extend(prompt_detections[int(i)] for i in flat_indices)

        final.sort(key=lambda det: (det.image, det.target, -det.confidence))
        return final

    def estimate_scale(
        self,
        image_name: str,
        detections: list[Detection],
        width: int,
        height: int,
        metadata: dict[str, Any],
    ) -> ScaleEstimate:
        relative_alt = metadata.get("relative_alt_m")
        focal_35mm = metadata.get("focal_length_35mm")
        gps_alt = metadata.get("gps_alt_m")

        altitude = relative_alt if relative_alt and relative_alt > 0 else None
        source = "DJI/XMP RelativeAltitude + EXIF 35mm"
        confidence = "Media"

        if altitude and focal_35mm and focal_35mm > 0:
            hfov = 2 * math.atan(36.0 / (2 * float(focal_35mm)))
            ground_width_m = 2 * float(altitude) * math.tan(hfov / 2)
            m_per_px = ground_width_m / max(1, width)
            return self.scale_with_bar(image_name, m_per_px, source, confidence, width)

        references = {
            "Persona": 1.70,
            "Cono vial": 0.70,
        }
        candidates = []
        for det in detections:
            if det.is_internal or det.target not in references or det.confidence < 0.35:
                continue
            if det.height_px < max(25, height * 0.025):
                continue
            candidates.append(references[det.target] / det.height_px)

        if candidates:
            m_per_px = float(np.median(candidates))
            return self.scale_with_bar(image_name, m_per_px, "Referencia visual persona/cono", "Baja", width)

        return ScaleEstimate(image_name, None, "Sin datos suficientes", "N/D", "Escala aprox.: N/D")

    def scale_with_bar(
        self,
        image_name: str,
        m_per_px: float,
        source: str,
        confidence: str,
        image_width: int,
    ) -> ScaleEstimate:
        target_m = max(m_per_px * image_width * 0.18, m_per_px * 80)
        bar_m = nice_distance(target_m)
        bar_px = int(round(bar_m / m_per_px)) if m_per_px > 0 else None
        if bar_px and bar_px > image_width * 0.35:
            bar_m = nice_distance(m_per_px * image_width * 0.25)
            bar_px = int(round(bar_m / m_per_px))
        if bar_px and bar_px < 35:
            bar_px = 35
            bar_m = bar_px * m_per_px
        label = f"Escala aprox.: {bar_m:g} m | {m_per_px * 100:.2f} cm/px | {confidence}"
        return ScaleEstimate(image_name, m_per_px, source, confidence, label, bar_m, bar_px)

    def evaluate_alerts(
        self,
        image_name: str,
        detections: list[Detection],
        width: int,
        height: int,
    ) -> list[Alert]:
        alerts: list[Alert] = []
        people = [det for det in detections if det.target == "Persona" and det.confidence >= max(0.25, self.confidence)]
        helmets = [det for det in detections if det.target == "Casco" and det.confidence >= max(0.18, self.confidence)]
        non_work_cues = [
            det
            for det in detections
            if det.target in {"Telefono celular", "Persona sentada", "Persona recostada", "Silla"} and det.confidence >= 0.30
        ]
        work_context = [
            det
            for det in detections
            if det.target
            in {
                "Chaleco reflectivo",
                "Cono vial",
                "Cartel de seguridad",
                "Camioneta",
                "Camion",
                "Excavadora",
                "Grua",
                "Caneria/Tuberia",
                "Zanja abierta",
                "Mojon/estaca",
                "Escalera",
                "Andamio",
            }
            and det.confidence >= 0.25
        ]

        for index, person in enumerate(people, start=1):
            has_helmet, helmet_conf = self.person_has_helmet(person, helmets, width, height)
            if not has_helmet:
                alerts.append(
                    Alert(
                        image=image_name,
                        tipo="Persona sin casco visible",
                        severidad="Alta",
                        confianza=round(float(person.confidence), 3),
                        detalle=f"Persona #{index} detectada sin casco asociado en la zona superior del cuerpo.",
                        accion_sugerida="Confirmar visibilidad de la cabeza y uso de EPP en la foto original o en campo.",
                        x1=person.x1,
                        y1=person.y1,
                        x2=person.x2,
                        y2=person.y2,
                        evidencia=f"person={person.confidence:.2f}; casco_asociado=0",
                    )
                )

            cue = self.strong_non_working_cue(person, non_work_cues, width, height)
            if cue and not self.near_work_context(person, work_context, helmets, width, height):
                confidence = min(0.99, (person.confidence + cue.confidence) / 2)
                if confidence >= 0.55:
                    alerts.append(
                        Alert(
                            image=image_name,
                            tipo="Postura u objeto para revision visual",
                            severidad="Media",
                            confianza=round(float(confidence), 3),
                            detalle=(
                                f"Persona #{index} con indicio fuerte: {cue.target}. "
                                "Una foto no permite inferir actividad o productividad laboral."
                            ),
                            accion_sugerida="Revisar contexto en campo; no clasificar rendimiento laboral a partir de la foto.",
                            x1=person.x1,
                            y1=person.y1,
                            x2=person.x2,
                            y2=person.y2,
                            evidencia=f"person={person.confidence:.2f}; {cue.target}={cue.confidence:.2f}",
                        )
                    )

        return alerts

    def person_has_helmet(
        self,
        person: Detection,
        helmets: list[Detection],
        width: int,
        height: int,
    ) -> tuple[bool, float]:
        person_box = detection_box(person)
        head_bottom = person.y1 + int(max(1, person.height_px) * 0.42)
        head_box = expand_box((person.x1, person.y1, person.x2, head_bottom), 0.18, 0.22, width, height)
        best = 0.0
        for helmet in helmets:
            helmet_box = detection_box(helmet)
            score = iou(head_box, helmet_box)
            if center_inside(helmet, head_box) or score > 0.01:
                best = max(best, helmet.confidence)
        return best > 0, best

    def strong_non_working_cue(
        self,
        person: Detection,
        cues: list[Detection],
        width: int,
        height: int,
    ) -> Detection | None:
        person_zone = expand_box(detection_box(person), 0.20, 0.12, width, height)
        best: Detection | None = None
        best_score = 0.0
        for cue in cues:
            cue_box = detection_box(cue)
            overlap = intersection_area(person_zone, cue_box) / max(1, min(box_area(person_zone), box_area(cue_box)))
            close = center_inside(cue, person_zone) or center_inside(person, cue_box)
            score = cue.confidence + overlap
            if (close or overlap > 0.10) and score > best_score:
                best = cue
                best_score = score
        return best

    def near_work_context(
        self,
        person: Detection,
        context: list[Detection],
        helmets: list[Detection],
        width: int,
        height: int,
    ) -> bool:
        zone = expand_box(detection_box(person), 0.85, 0.45, width, height)
        if self.person_has_helmet(person, helmets, width, height)[0]:
            return True
        for det in context:
            if det.target == "Persona":
                continue
            det_box = detection_box(det)
            if center_inside(det, zone) or intersection_area(zone, det_box) > 0:
                return True
        return False

    def draw_detections(
        self,
        image: np.ndarray,
        image_name: str,
        detections: list[Detection],
        alerts: list[Alert],
        scale: ScaleEstimate,
    ) -> np.ndarray:
        height, width = image.shape[:2]
        thickness = max(2, int(round(max(width, height) / 900)))
        font_scale = max(0.55, min(1.25, max(width, height) / 1800))
        target_by_label = {target.label: target for target in self.targets}

        visible_count = sum(1 for det in detections if not det.is_internal)
        title = f"{image_name} | Hallazgos: {visible_count} | Alertas: {len(alerts)}"

        for detection in detections:
            if detection.is_internal:
                continue
            target = target_by_label.get(detection.target)
            color = hex_to_bgr(target.color if target else "#2563eb")
            cv2.rectangle(image, (detection.x1, detection.y1), (detection.x2, detection.y2), color, thickness)
            label = f"{detection.target} {detection.confidence:.2f}"
            (label_width, label_height), baseline = cv2.getTextSize(
                label,
                cv2.FONT_HERSHEY_SIMPLEX,
                font_scale,
                thickness,
            )
            x1 = detection.x1
            y1 = max(label_height + baseline + 6, detection.y1)
            cv2.rectangle(
                image,
                (x1, y1 - label_height - baseline - 8),
                (min(width - 1, x1 + label_width + 10), y1 + 4),
                color,
                -1,
            )
            cv2.putText(
                image,
                label,
                (x1 + 5, y1 - baseline - 2),
                cv2.FONT_HERSHEY_SIMPLEX,
                font_scale,
                (255, 255, 255),
                max(1, thickness - 1),
                cv2.LINE_AA,
            )

        for alert in alerts:
            if None in (alert.x1, alert.y1, alert.x2, alert.y2):
                continue
            x1, y1, x2, y2 = int(alert.x1), int(alert.y1), int(alert.x2), int(alert.y2)
            color = (36, 36, 230)
            cv2.rectangle(image, (x1, y1), (x2, y2), color, thickness + 2)
            alert_label = f"ALERTA: {alert.tipo}"
            (label_width, label_height), baseline = cv2.getTextSize(
                alert_label,
                cv2.FONT_HERSHEY_SIMPLEX,
                font_scale,
                thickness,
            )
            label_y = min(height - 4, max(label_height + baseline + 8, y2 + label_height + 12))
            cv2.rectangle(
                image,
                (x1, label_y - label_height - baseline - 8),
                (min(width - 1, x1 + label_width + 10), label_y + 4),
                color,
                -1,
            )
            cv2.putText(
                image,
                alert_label,
                (x1 + 5, label_y - baseline - 2),
                cv2.FONT_HERSHEY_SIMPLEX,
                font_scale,
                (255, 255, 255),
                max(1, thickness - 1),
                cv2.LINE_AA,
            )

        self.draw_scale_bar(image, scale)
        # Put the report heading outside the source pixels, clear of object labels.
        header_h = max(46, int(58 * font_scale))
        image = cv2.copyMakeBorder(image, header_h, 0, 0, 0, cv2.BORDER_CONSTANT, value=(20, 34, 50))
        title = title[:90]
        text_w = cv2.getTextSize(title, cv2.FONT_HERSHEY_SIMPLEX, font_scale, thickness)[0][0]
        title_scale = font_scale * min(1, max(1, width-24)/max(1, text_w))
        cv2.putText(image, title, (12, int(header_h*.65)), cv2.FONT_HERSHEY_SIMPLEX,
                    title_scale, (255, 255, 255), max(1, thickness-1), cv2.LINE_AA)
        return image

    def draw_scale_bar(self, image: np.ndarray, scale: ScaleEstimate) -> None:
        height, width = image.shape[:2]
        margin = max(14, int(width * 0.018))
        panel_h = max(52, int(height * 0.055))
        x = margin
        y = height - margin - panel_h
        panel_w = min(width - 2 * margin, max(360, int(width * 0.34)))
        overlay = image.copy()
        cv2.rectangle(overlay, (x, y), (x + panel_w, y + panel_h), (15, 23, 42), -1)
        cv2.addWeighted(overlay, 0.80, image, 0.20, 0, image)

        font_scale = max(0.48, min(0.85, width / 2200))
        thickness = max(1, int(round(width / 1500)))
        text_y = y + int(panel_h * 0.38)
        cv2.putText(
            image,
            scale.label[:95],
            (x + 12, text_y),
            cv2.FONT_HERSHEY_SIMPLEX,
            font_scale,
            (255, 255, 255),
            thickness,
            cv2.LINE_AA,
        )

        if scale.bar_px and scale.bar_m:
            bar_x = x + 14
            bar_y = y + int(panel_h * 0.72)
            bar_px = min(scale.bar_px, panel_w - 35)
            cv2.line(image, (bar_x, bar_y), (bar_x + bar_px, bar_y), (255, 255, 255), thickness + 3)
            cv2.line(image, (bar_x, bar_y - 8), (bar_x, bar_y + 8), (255, 255, 255), thickness + 2)
            cv2.line(image, (bar_x + bar_px, bar_y - 8), (bar_x + bar_px, bar_y + 8), (255, 255, 255), thickness + 2)
            cv2.putText(
                image,
                f"{bar_px * scale.m_per_px:g} m",
                (bar_x + bar_px + 10, bar_y + 6),
                cv2.FONT_HERSHEY_SIMPLEX,
                font_scale,
                (255, 255, 255),
                thickness,
                cv2.LINE_AA,
            )

    def detections_dataframe(self) -> pd.DataFrame:
        if not self.detections:
            return pd.DataFrame(
                columns=[
                    "image",
                    "target",
                    "group",
                    "priority",
                    "confidence",
                    "x1",
                    "y1",
                    "x2",
                    "y2",
                    "source_pass",
                    "is_internal",
                ]
            )
        return pd.DataFrame(asdict(det) for det in self.detections)

    def alerts_dataframe(self) -> pd.DataFrame:
        if not self.alerts:
            return pd.DataFrame(
                columns=[
                    "image",
                    "tipo",
                    "severidad",
                    "confianza",
                    "detalle",
                    "accion_sugerida",
                    "x1",
                    "y1",
                    "x2",
                    "y2",
                    "evidencia",
                ]
            )
        return pd.DataFrame(asdict(alert) for alert in self.alerts)

    def alert_summary_dataframe(self) -> pd.DataFrame:
        df = self.alerts_dataframe()
        if df.empty:
            return pd.DataFrame(
                [
                    {
                        "severidad": "N/D",
                        "tipo": "Sin alertas automaticas (no certifica seguridad)",
                        "cantidad": 0,
                        "fotos_afectadas": 0,
                        "confianza_maxima": "",
                        "accion_sugerida": "No se generaron alertas automaticas.",
                    }
                ]
            )

        rows = []
        for (severity, alert_type), group in df.groupby(["severidad", "tipo"]):
            rows.append(
                {
                    "severidad": severity,
                    "tipo": alert_type,
                    "cantidad": int(len(group)),
                    "fotos_afectadas": int(group["image"].nunique()),
                    "confianza_maxima": round(float(group["confianza"].max()), 3),
                    "accion_sugerida": str(group["accion_sugerida"].iloc[0]),
                }
            )
        result = pd.DataFrame(rows)
        severity_order = {"Alta": 0, "Media": 1, "Baja": 2, "OK": 9}
        result["_order"] = result["severidad"].map(severity_order).fillna(8)
        return result.sort_values(["_order", "cantidad"], ascending=[True, False]).drop(columns=["_order"])

    def images_dataframe(self) -> pd.DataFrame:
        if not self.image_summaries:
            return pd.DataFrame()
        return pd.DataFrame(asdict(summary) for summary in self.image_summaries)

    def encounters_dataframe(self) -> pd.DataFrame:
        df = self.detections_dataframe()
        if not df.empty and "is_internal" in df.columns:
            df = df[df["is_internal"] == False]  # noqa: E712
        if df.empty:
            return pd.DataFrame(
                columns=[
                    "hallazgo",
                    "cantidad",
                    "imagenes_con_hallazgo",
                    "confianza_promedio",
                    "confianza_maxima",
                    "grupo",
                    "prioridad",
                    "accion_sugerida",
                ]
            )

        target_action = {target.label: target.action for target in self.targets}
        rows = []
        for target, group in df.groupby("target"):
            rows.append(
                {
                    "hallazgo": target,
                    "cantidad": int(len(group)),
                    "imagenes_con_hallazgo": int(group["image"].nunique()),
                    "confianza_promedio": round(float(group["confidence"].mean()), 3),
                    "confianza_maxima": round(float(group["confidence"].max()), 3),
                    "grupo": str(group["group"].iloc[0]),
                    "prioridad": str(group["priority"].iloc[0]),
                    "accion_sugerida": target_action.get(target, "Validar visualmente y registrar decision."),
                }
            )
        result = pd.DataFrame(rows)
        priority_order = {"Alta": 0, "Media": 1, "Baja": 2}
        result["_priority_order"] = result["prioridad"].map(priority_order).fillna(9)
        result = result.sort_values(["_priority_order", "cantidad", "hallazgo"], ascending=[True, False, True])
        return result.drop(columns=["_priority_order"])

    def config_dataframe(self) -> pd.DataFrame:
        rows = [
            ("fecha_inicio", self.started_at.strftime("%Y-%m-%d %H:%M:%S")),
            ("fecha_fin", self.finished_at.strftime("%Y-%m-%d %H:%M:%S") if self.finished_at else ""),
            ("input_folder", str(self.input_folder)),
            ("output_folder", str(self.output_folder)),
            ("modelo", self.model_name),
            ("control_syh", self.safety_control),
            ("indicios_postura", self.activity_cues),
            ("coordenadas_cajas", "pixeles de imagen procesada, EXIF orientado; extremos x2/y2 exclusivos"),
            ("confianza", self.confidence),
            ("perfil_busqueda", self.search_profile),
            ("pasadas", ", ".join(search_pass.name for search_pass in self.search_passes)),
            ("nms_iou", self.nms_iou),
            ("max_image_side", self.max_image_side),
            ("batch_size", self.batch_size),
            ("imgsz", self.imgsz),
            ("objetivos", ", ".join(f"{target.label} -> {target.prompt}" for target in self.targets)),
        ]
        return pd.DataFrame(rows, columns=["parametro", "valor"])

    def export_all(self) -> dict[str, Path]:
        artifacts = {
            "alertas_csv": self.write_csv(self.alerts_dataframe(), "alertas_detalle.csv"),
            "alertas_resumen_csv": self.write_csv(self.alert_summary_dataframe(), "alertas_resumen.csv"),
            "detecciones_csv": self.write_csv(self.detections_dataframe(), "detecciones_detalle.csv"),
            "imagenes_csv": self.write_csv(self.images_dataframe(), "imagenes_resumen.csv"),
            "encuentros_csv": self.write_csv(self.encounters_dataframe(), "encuentros_resumen.csv"),
            "excel": self.write_excel(),
            "pdf": self.write_pdf(),
        }
        artifacts["zip"] = self.write_zip(artifacts)
        return artifacts

    def write_csv(self, dataframe: pd.DataFrame, filename: str) -> Path:
        path = self.tables_folder / filename
        dataframe.to_csv(path, index=False, encoding="utf-8-sig")
        return path

    def write_excel(self) -> Path:
        path = self.output_folder / "reporte_inspeccion_visual.xlsx"
        with pd.ExcelWriter(path, engine="openpyxl") as writer:
            self.alert_summary_dataframe().to_excel(writer, index=False, sheet_name="Resumen_alertas")
            self.alerts_dataframe().to_excel(writer, index=False, sheet_name="Alertas_detalle")
            self.encounters_dataframe().to_excel(writer, index=False, sheet_name="Encuentros")
            self.detections_dataframe().to_excel(writer, index=False, sheet_name="Detalle_detecciones")
            self.images_dataframe().to_excel(writer, index=False, sheet_name="Fotos")
            self.config_dataframe().to_excel(writer, index=False, sheet_name="Configuracion")
            self.recommendations_dataframe().to_excel(writer, index=False, sheet_name="Recomendaciones")

        self.style_excel(path)
        return path

    def style_excel(self, path: Path) -> None:
        try:
            from openpyxl import load_workbook
            from openpyxl.styles import Alignment, Font, PatternFill
        except Exception:
            return

        workbook = load_workbook(path)
        header_fill = PatternFill("solid", fgColor="1F4E5F")
        header_font = Font(color="FFFFFF", bold=True)
        soft_fill = PatternFill("solid", fgColor="EAF4F4")

        for sheet in workbook.worksheets:
            sheet.freeze_panes = "A2"
            for cell in sheet[1]:
                cell.fill = header_fill
                cell.font = header_font
                cell.alignment = Alignment(horizontal="center", vertical="center")
            for row in sheet.iter_rows(min_row=2):
                for cell in row:
                    cell.alignment = Alignment(vertical="top", wrap_text=True)
                if row[0].row % 2 == 0:
                    for cell in row:
                        cell.fill = soft_fill
            for column_cells in sheet.columns:
                column = column_cells[0].column_letter
                max_length = max(len(str(cell.value or "")) for cell in column_cells[:80])
                sheet.column_dimensions[column].width = min(max(12, max_length + 2), 48)

        workbook.save(path)

    def recommendations_dataframe(self) -> pd.DataFrame:
        encounters = self.encounters_dataframe()
        if encounters.empty:
            return pd.DataFrame(
                [
                    {
                        "tema": "Resultado",
                        "grupo": "Gestion",
                        "prioridad": "Media",
                        "cantidad": "",
                        "recomendacion": "No se detectaron los objetivos configurados. Revisar muestra y sensibilidad si era esperable encontrarlos.",
                    }
                ]
            )

        rows = []
        for _, row in encounters.iterrows():
            rows.append(
                {
                    "tema": row["hallazgo"],
                    "grupo": row["grupo"],
                    "prioridad": row["prioridad"],
                    "cantidad": row["cantidad"],
                    "recomendacion": row["accion_sugerida"],
                }
            )
        rows.append(
            {
                "tema": "Uso del reporte",
                "grupo": "Gestion",
                "prioridad": "Media",
                "cantidad": "",
                "recomendacion": "La IA acelera el relevamiento, pero cada hallazgo debe validarse por responsable tecnico o SyH antes de emitir conformidad.",
            }
        )
        return pd.DataFrame(rows)

    def write_pdf(self) -> Path:
        try:
            from reportlab.lib import colors
            from reportlab.lib.enums import TA_CENTER
            from reportlab.lib.pagesizes import A4, landscape
            from reportlab.lib.styles import ParagraphStyle, getSampleStyleSheet
            from reportlab.lib.units import mm
            from reportlab.platypus import PageBreak, Paragraph, SimpleDocTemplate, Spacer, Table, TableStyle
        except ImportError as exc:
            raise RuntimeError("Falta reportlab. Instala con: pip install reportlab") from exc

        pdf_path = self.output_folder / "reporte_inspeccion_visual.pdf"
        doc = SimpleDocTemplate(
            str(pdf_path),
            pagesize=landscape(A4),
            leftMargin=12 * mm,
            rightMargin=12 * mm,
            topMargin=12 * mm,
            bottomMargin=10 * mm,
        )

        styles = getSampleStyleSheet()
        styles.add(
            ParagraphStyle(
                "ReportTitle",
                parent=styles["Title"],
                fontName="Helvetica-Bold",
                fontSize=24,
                textColor=colors.HexColor("#16323F"),
                alignment=TA_CENTER,
                spaceAfter=10,
            )
        )
        styles.add(
            ParagraphStyle(
                "SectionTitle",
                parent=styles["Heading2"],
                fontSize=14,
                textColor=colors.HexColor("#1F4E5F"),
                spaceBefore=8,
                spaceAfter=8,
            )
        )
        styles.add(
            ParagraphStyle(
                "Small",
                parent=styles["Normal"],
                fontSize=8,
                leading=10,
                textColor=colors.HexColor("#334155"),
            )
        )

        story: list[Any] = []
        story.append(Paragraph("Reporte de Inspeccion Visual", styles["ReportTitle"]))
        story.append(
            Paragraph(
                "Oficina tecnica + Seguridad e Higiene | Deteccion multiobjeto con IA | "
                f"{datetime.now().strftime('%d/%m/%Y %H:%M')}",
                styles["Normal"],
            )
        )
        story.append(Spacer(1, 10))

        total_images = len(self.image_summaries)
        images_with_findings = sum(1 for summary in self.image_summaries if summary.total_detections > 0)
        total_detections = sum(1 for det in self.detections if not det.is_internal)
        total_alerts = len(self.alerts)
        summary_data = [
            ["Fotos procesadas", "Fotos con hallazgos", "Alertas", "Detecciones", "Perfil"],
            [
                str(total_images),
                str(images_with_findings),
                str(total_alerts),
                str(total_detections),
                self.search_profile,
            ],
        ]
        summary_table = Table(summary_data, colWidths=[38 * mm, 42 * mm, 35 * mm, 42 * mm, 42 * mm])
        summary_table.setStyle(
            TableStyle(
                [
                    ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#1F4E5F")),
                    ("TEXTCOLOR", (0, 0), (-1, 0), colors.white),
                    ("FONTNAME", (0, 0), (-1, 0), "Helvetica-Bold"),
                    ("ALIGN", (0, 0), (-1, -1), "CENTER"),
                    ("GRID", (0, 0), (-1, -1), 0.4, colors.HexColor("#CBD5E1")),
                    ("BACKGROUND", (0, 1), (-1, 1), colors.HexColor("#EAF4F4")),
                    ("FONTSIZE", (0, 0), (-1, -1), 10),
                ]
            )
        )
        story.append(summary_table)

        story.append(Paragraph("Resumen de alertas", styles["SectionTitle"]))
        alert_summary = self.alert_summary_dataframe()
        alert_rows = [["Severidad", "Tipo", "Cantidad", "Fotos", "Conf. max.", "Accion sugerida"]]
        for _, row in alert_summary.iterrows():
            alert_rows.append(
                [
                    html.escape(str(row["severidad"])),
                    html.escape(str(row["tipo"])),
                    str(row["cantidad"]),
                    str(row["fotos_afectadas"]),
                    str(row["confianza_maxima"]),
                    Paragraph(html.escape(str(row["accion_sugerida"])), styles["Small"]),
                ]
            )
        alert_table = Table(alert_rows, colWidths=[24 * mm, 52 * mm, 20 * mm, 20 * mm, 24 * mm, 76 * mm])
        alert_table.setStyle(self.default_table_style(colors))
        story.append(alert_table)

        story.append(Paragraph("Encuentros y cantidades", styles["SectionTitle"]))
        encounters = self.encounters_dataframe()
        if encounters.empty:
            story.append(Paragraph("No se detectaron los objetivos configurados.", styles["Normal"]))
        else:
            encounter_rows = [["Hallazgo", "Cantidad", "Fotos", "Conf. prom.", "Grupo", "Prioridad"]]
            for _, row in encounters.iterrows():
                encounter_rows.append(
                    [
                        html.escape(str(row["hallazgo"])),
                        str(row["cantidad"]),
                        str(row["imagenes_con_hallazgo"]),
                        str(row["confianza_promedio"]),
                        html.escape(str(row["grupo"])),
                        html.escape(str(row["prioridad"])),
                    ]
                )
            table = Table(encounter_rows, colWidths=[52 * mm, 24 * mm, 22 * mm, 28 * mm, 42 * mm, 30 * mm])
            table.setStyle(self.default_table_style(colors))
            story.append(table)

        story.append(Paragraph("Lectura tecnica sugerida", styles["SectionTitle"]))
        recs = self.recommendations_dataframe().head(8)
        rec_rows = [["Tema", "Prioridad", "Recomendacion"]]
        for _, row in recs.iterrows():
            rec_rows.append(
                [
                    html.escape(str(row.get("tema", ""))),
                    html.escape(str(row.get("prioridad", ""))),
                    Paragraph(html.escape(str(row.get("recomendacion", ""))), styles["Small"]),
                ]
            )
        rec_table = Table(rec_rows, colWidths=[48 * mm, 27 * mm, 122 * mm])
        rec_table.setStyle(self.default_table_style(colors))
        story.append(rec_table)

        story.append(PageBreak())
        story.append(Paragraph("Detalle por foto", styles["SectionTitle"]))
        photo_rows = [["Foto", "Estado", "Alertas", "Hallazgos", "Escala / GPS"]]
        for summary in self.image_summaries:
            gps = ""
            if summary.gps_lat is not None and summary.gps_lon is not None:
                gps = f"{summary.gps_lat:.6f}, {summary.gps_lon:.6f}"
            if summary.gps_alt_m is not None:
                gps = (gps + " | " if gps else "") + f"{summary.gps_alt_m:.1f} m"
            if summary.scale_m_per_px is not None:
                gps = (f"Aprox. {summary.scale_m_per_px * 100:.2f} cm/px | " + gps).strip(" |")
            photo_rows.append(
                [
                    html.escape(summary.image[:48]),
                    summary.status,
                    str(summary.alert_count),
                    Paragraph(html.escape(summary.findings), styles["Small"]),
                    gps or "N/D",
                ]
            )
        photo_table = Table(photo_rows, colWidths=[52 * mm, 30 * mm, 18 * mm, 78 * mm, 44 * mm], repeatRows=1)
        photo_table.setStyle(self.default_table_style(colors))
        story.append(photo_table)

        annotated = [summary for summary in self.image_summaries if summary.annotated_path]
        if annotated:
            story.append(PageBreak())
            story.append(Paragraph("Anexo fotografico", styles["SectionTitle"]))
            for summary in annotated:
                story.append(Paragraph(html.escape(summary.image), styles["Heading3"]))
                story.append(Paragraph(html.escape(summary.findings), styles["Small"]))
                image_flowable = self.reportlab_image(summary.annotated_path, max_width=240 * mm, max_height=128 * mm)
                if image_flowable:
                    story.append(image_flowable)
                    story.append(Spacer(1, 8))

        story.append(Spacer(1, 6))
        story.append(
            Paragraph(
                "Nota: este reporte es una ayuda de deteccion y trazabilidad. No reemplaza la inspeccion de un "
                "profesional habilitado ni el criterio del responsable de Higiene y Seguridad.",
                styles["Small"],
            )
        )

        def page_footer(canvas: Any, doc_obj: Any) -> None:
            canvas.saveState()
            canvas.setFont("Helvetica", 7)
            canvas.setFillColor(colors.HexColor("#64748B"))
            canvas.drawRightString(285 * mm, 7 * mm, f"Pagina {doc_obj.page}")
            canvas.drawString(12 * mm, 7 * mm, "Reporte generado por Buscador Visual Oficina Tecnica")
            canvas.restoreState()

        doc.build(story, onFirstPage=page_footer, onLaterPages=page_footer)
        return pdf_path

    @staticmethod
    def default_table_style(colors_module: Any) -> Any:
        from reportlab.platypus import TableStyle

        return TableStyle(
            [
                ("BACKGROUND", (0, 0), (-1, 0), colors_module.HexColor("#1F4E5F")),
                ("TEXTCOLOR", (0, 0), (-1, 0), colors_module.white),
                ("FONTNAME", (0, 0), (-1, 0), "Helvetica-Bold"),
                ("ALIGN", (0, 0), (-1, 0), "CENTER"),
                ("VALIGN", (0, 0), (-1, -1), "TOP"),
                ("GRID", (0, 0), (-1, -1), 0.35, colors_module.HexColor("#CBD5E1")),
                ("ROWBACKGROUNDS", (0, 1), (-1, -1), [colors_module.white, colors_module.HexColor("#F8FAFC")]),
                ("FONTSIZE", (0, 0), (-1, -1), 8),
                ("LEADING", (0, 0), (-1, -1), 9),
            ]
        )

    @staticmethod
    def reportlab_image(path: str | Path, max_width: float, max_height: float) -> Any | None:
        try:
            from reportlab.platypus import Image as RLImage

            with Image.open(path) as image:
                width, height = image.size
            scale = min(max_width / width, max_height / height)
            return RLImage(str(path), width=width * scale, height=height * scale)
        except Exception:
            return None

    def write_zip(self, artifacts: dict[str, Path]) -> Path:
        zip_path = self.output_folder / "paquete_reporte_visual.zip"
        with zipfile.ZipFile(zip_path, "w", compression=zipfile.ZIP_DEFLATED) as zf:
            for name, path in artifacts.items():
                if name == "zip" or not Path(path).exists():
                    continue
                zf.write(path, arcname=Path(path).relative_to(self.output_folder))
            for image_path in sorted(self.annotated_folder.glob("*.jpg")):
                zf.write(image_path, arcname=image_path.relative_to(self.output_folder))
        return zip_path


def run_inspection(
    input_folder: str | Path,
    output_folder: str | Path = "output_reporte_visual",
    preset_labels: Iterable[str] | None = DEFAULT_PRESET_LABELS,
    extra_terms: str | Iterable[str] | None = None,
    confidence: float = 0.20,
    search_profile: str = "equilibrada",
    batch_size: int = 8,
    model_name: str = "yolov8s-worldv2.pt",
) -> tuple[VisualInspectionTool, dict[str, Path]]:
    targets = build_targets(preset_labels=preset_labels, extra_terms=extra_terms)
    tool = VisualInspectionTool(
        input_folder=input_folder,
        output_folder=output_folder,
        targets=targets,
        confidence=confidence,
        search_profile=search_profile,
        batch_size=batch_size,
        model_name=model_name,
    )
    tool.process()
    artifacts = tool.export_all()
    return tool, artifacts


def copy_uploaded_files(uploaded: dict[str, bytes], input_folder: str | Path) -> None:
    folder = file_path(input_folder)
    folder.mkdir(parents=True, exist_ok=True)
    for filename, content in uploaded.items():
        name = Path(filename.replace("\\", "/")).name
        if Path(name).suffix.lower() in IMAGE_EXTENSIONS:
            destination = folder / name
            if destination.exists():
                raise FileExistsError(f"Ya existe {destination}; renombrar la nueva imagen.")
            destination.write_bytes(content)


def _running_in_notebook() -> bool:
    return "ipykernel" in sys.modules or "google.colab" in sys.modules


def main(argv: list[str] | None = None) -> int:
    parser = argparse.ArgumentParser(description="Buscador visual multiobjeto para reportes de obra.")
    parser.add_argument("--input", required=True, help="Carpeta con fotos.")
    parser.add_argument("--output", default="output_reporte_visual", help="Carpeta de salida.")
    parser.add_argument("--buscar", default="", help="Objetos extra separados por coma: cono, cartel, casco, autos.")
    parser.add_argument(
        "--preset",
        action="append",
        default=[],
        help="Etiqueta de catalogo a incluir. Puede repetirse. Si se omite, usa el preset recomendado.",
    )
    parser.add_argument("--confianza", type=float, default=0.20, help="Confianza minima de deteccion.")
    parser.add_argument(
        "--perfil",
        choices=sorted(SEARCH_PROFILES),
        default="equilibrada",
        help="rapida, equilibrada o profunda.",
    )
    parser.add_argument("--batch-size", type=int, default=8, help="Cantidad de sectores por lote.")
    parser.add_argument("--modelo", default="yolov8s-worldv2.pt", help="Modelo YOLO-World.")
    args = parser.parse_args(argv)

    presets = args.preset or list(DEFAULT_PRESET_LABELS)
    _, artifacts = run_inspection(
        input_folder=args.input,
        output_folder=args.output,
        preset_labels=presets,
        extra_terms=args.buscar,
        confidence=args.confianza,
        search_profile=args.perfil,
        batch_size=args.batch_size,
        model_name=args.modelo,
    )
    print(json.dumps({key: str(path) for key, path in artifacts.items()}, indent=2, ensure_ascii=False))
    return 0

```

## Interfaz

```python
"""Notebook interface; the engine cell must have been executed first."""
from pathlib import Path
import traceback
import ipywidgets as widgets
from IPython.display import display, clear_output, FileLink

try:
    from google.colab import files as colab_files
except ImportError:
    colab_files = None

entrada = widgets.Text(value=str(Path.cwd()/"fotos"), description="Fotos", layout=widgets.Layout(width="98%"))
destino = widgets.Text(value=str(Path.cwd()/"resultados"), description="Salida", layout=widgets.Layout(width="98%"))
objetivos = widgets.SelectMultiple(options=[t.label for t in CATALOG], value=DEFAULT_PRESET_LABELS,
                                  description="Objetivos", layout=widgets.Layout(width="98%", height="210px"))
extras = widgets.Text(value="", description="Extra", placeholder="Otros objetos; preferir prompts en ingles", layout=widgets.Layout(width="98%"))
perfil = widgets.Dropdown(options=[("Rapida", "rapida"), ("Equilibrada", "equilibrada"), ("Profunda", "profunda")], value="equilibrada", description="Perfil")
confianza = widgets.FloatSlider(value=.20, min=.01, max=.9, step=.01, description="Confianza")
lote = widgets.IntSlider(value=4, min=1, max=16, description="Lote")
modelo = widgets.Dropdown(options=["yolov8s-worldv2.pt", "yolov8m-worldv2.pt", "yolov8l-worldv2.pt"], description="Modelo")
syh = widgets.Checkbox(value=True, description="Control persona y casco")
posturas = widgets.Checkbox(value=False, description="Indicios de postura/objetos para revision visual")
subir = widgets.Button(description="Subir fotos (Colab)", button_style="info", disabled=colab_files is None)
procesar = widgets.Button(description="Procesar fotos", button_style="success")
descargar = widgets.Button(description="Descargar ZIP", disabled=True)
salida = widgets.Output()
artefactos = {}
_modelo_cache = {}


def cargar_fotos(_):
    with salida:
        clear_output()
        try:
            copy_uploaded_files(colab_files.upload(), file_path(entrada.value))
            print(f"Fotos disponibles: {len(image_files(file_path(entrada.value)))}")
        except (OSError, ValueError) as exc:
            print(str(exc))


def ejecutar_industrial(_):
    global artefactos
    procesar.disabled = True
    descargar.disabled = True
    artefactos = {}
    with salida:
        clear_output()
        try:
            targets = build_targets(objetivos.value, extras.value)
            if not targets:
                raise ValueError("Selecciona al menos un objetivo.")
            tool = VisualInspectionTool(entrada.value, destino.value, targets=targets,
                                        confidence=confianza.value, search_profile=perfil.value,
                                        batch_size=lote.value, model_name=modelo.value,
                                        safety_control=syh.value, activity_cues=posturas.value)
            key = (tool.model_name, tuple(tool.prompts))
            if key in _modelo_cache:
                tool.model = _modelo_cache[key]
            else:
                _modelo_cache.clear()
            tool.process()
            _modelo_cache[key] = tool.model
            artefactos = tool.export_all()
            for name in ("pdf", "excel", "zip"):
                display(FileLink(str(artefactos[name])))
            descargar.disabled = False
        except Exception:
            traceback.print_exc()
        finally:
            procesar.disabled = False


def descargar_industrial(_):
    path = artefactos.get("zip")
    if path and colab_files is not None:
        colab_files.download(str(path))
    elif path:
        with salida:
            print(f"ZIP: {path}")


subir.on_click(cargar_fotos)
procesar.on_click(ejecutar_industrial)
descargar.on_click(descargar_industrial)
display(widgets.HTML("<h3>Buscador industrial</h3><p>Seleccioná fotos, objetivos y perfil. Cada ejecución crea una carpeta nueva.</p>"))
display(entrada, destino, objetivos, extras, perfil, confianza, lote, modelo, syh, posturas)
display(widgets.HBox([subir, procesar, descargar]), salida)
```

## Uso y límites

- Ejecutar instalación, motor e interfaz, en ese orden. Elegir carpeta y objetivos; pulsar **Procesar fotos**.
- En Colab, usar **Subir fotos**. Localmente, copiar fotos a la ruta indicada.
- Se conserva resolución original para los sectores. Las fotos completas deben caber en RAM; para ortomosaicos usar la versión agro.
- La escala visual es aproximada y dependiente de perspectiva. La altitud GPS no se usa como altura sobre terreno.
- Ausencia de casco detectado no demuestra ausencia de casco real. Los indicios de postura son opcionales y no permiten inferir productividad.
- Los conteos son por foto; un mismo objeto visto en dos fotos se cuenta en ambas.
- Verificación realizada con imágenes y detecciones sintéticas; falta validar el modelo en fotografías reales.
